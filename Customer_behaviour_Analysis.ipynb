```python
# fill missing values in 'Review Rating' column with the median rating of respective 'Category'
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

df.groupby('Category') - Groups the DataFrame by the 'Category' column (e.g., Electronics, Clothing, Books). Each unique category becomes a separate group.​

['Review Rating'] - Selects only the 'Review Rating' column from each group.​

.transform() - Applies a function to each group and returns a result with the same shape and index as the original DataFrame. Unlike aggregation (which reduces data), transform maintains the original structure.​

lambda x: x.fillna(x.median()) - This is an anonymous function that:

Takes each group (x) of Review Ratings

Calculates the median for that specific group using x.median()

Fills any missing values (NaN) in that group with the group's median using x.fillna()​

df['Review Rating'] = - Replaces the original 'Review Rating' column with the transformed version (now with filled values).​

Practical Example
If you have:

Electronics: Ratings [5, NaN, 4, 3] → median = 4 → fills NaN with 4

Clothing: Ratings [2, NaN, 3, 1] → median = 2.5 → fills NaN with 2.5

This approach is better than filling with the overall median because it preserves category-specific patterns. Electronics might have higher ratings than Clothing, so using category-specific medians keeps the data more accurate.

```python
#to create a new column with name age_group
Labels = ['young Adult', 'Adult','Middle-aged','Senior']
df ['age_group'] = pd.qcut(df['age'], q=4, labels=Labels)
```

labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
 Creates a list of 4 custom labels that will be assigned to each age group.

 df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)
Creates a new column called 'age_group' by dividing the 'age' column into 4 quantile-based bins.

df['age'] - The numeric column to be categorized.​

q=4 - Divides the data into 4 quartiles (25% of people in each group). Unlike regular binning, qcut ensures equal counts in each bin, not equal size ranges.​

labels = labels - Assigns the custom labels to each quartile in order.​

How It Works
If you have ages: 18, 22, 25, 30, 35, 40, 45, 50, 55, 60, 65, 70

The function will:

Sort all ages

Divide them into 4 groups with 3 people each (equal counts)​

Assign labels: lowest 25% = "Young Adult", next 25% = "Adult", next 25% = "Middle-aged", highest 25% = "Senior"​

Key difference from cut(): pd.qcut() creates bins with equal frequencies (same number of people per group), while pd.cut() creates bins with equal width (same age range per group). This is useful when your age distribution is skewed

```python
# create new column purchase_frequency_days
frequency_mapping = {
        'Fortnightly': 14,
        'Weekly': 7,
        'Monthly': 30,
        'Quarterly': 90,
        'Bi-Weekly': 14,
        'Annually': 365,
        'Every 3 Months': 90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

df['purchase_frequency_days'] - Creates a new column in your DataFrame called 'purchase_frequency_days'.

df['frequency_of_purchases'] - Selects the existing column that contains purchase frequency information (likely text values like "Weekly", "Monthly", "Quarterly").

.map(frequency_mapping) - Applies a mapping transformation using a dictionary called frequency_mapping. This function takes each value in the column and replaces it with the corresponding value from the dictionary.

The map() function operates element-wise, meaning it processes one value at a time. It accepts three types of arguments: a function, a dictionary, or a Series. In your case, it's using a dictionary.
